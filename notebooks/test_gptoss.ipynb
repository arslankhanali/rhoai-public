# GPT check


```python
import json
import os
import requests

url = os.environ.get(
    "PREDICTOR_URL",
    "http://gpt-oss-20b-predictor.model-serving.svc.cluster.local:8080/v1/chat",
)
body = {
    "model": os.environ.get("INFERENCE_MODEL", "gpt-oss-20b"),  
    "prompt": "User: Hi\nAssistant:",
    "max_tokens": 32,
}
h = {"Content-Type": "application/json"}
tok = os.environ.get("INFERENCE_BEARER_TOKEN") or os.environ.get("OC_TOKEN")
if tok:
    h["Authorization"] = f"Bearer {tok}"
r = requests.post(url, data=json.dumps(body), headers=h, timeout=120, verify=False)
print(r.status_code, r.text[:1500])
```
