```python
import os, requests

url = os.environ.get(
    "PREDICTOR_URL",
    "http://gpt-oss-20b-predictor.canberra.svc.cluster.local:8080/v1/chat/completions",
)

body = {
    "model": os.environ.get("INFERENCE_MODEL", "gpt-oss-20b"),
    "messages": [
        {"role": "user", "content": "Hi"}
    ],
    "max_tokens": 32,
    "temperature": 0.2,
}

h = {"Content-Type": "application/json"}
tok = os.environ.get("INFERENCE_BEARER_TOKEN") or os.environ.get("OC_TOKEN")
if tok:
    h["Authorization"] = f"Bearer {tok}"

r = requests.post(url, json=body, headers=h, timeout=120, verify=False)
print(r.status_code, r.text[:1500])
```
